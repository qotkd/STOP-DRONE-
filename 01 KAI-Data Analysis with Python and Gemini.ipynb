1. 원의 면적을 계산하는 프로그램을 작성해 줘. 계산 프로그램의 결과는 소수 첫째 아래 자리에서 반올림해 줘. 즉, 실행창에서 3을 입력하여 실행하면, 다음과 같이 실행창에 출력해야 한다.
  ==> 반지름이 3인 원의 넓이는 28.3.


```python
import math

# Get radius input from the user
radius_str = input("반지름을 입력하세요: ")
radius = float(radius_str)

# Calculate the area of the circle
area = math.pi * radius * radius

# Round the area to the first decimal place
rounded_area = round(area, 1)

# Print the result in the specified format
print(f"반지름이 {int(radius)}인 원의 넓이는 {rounded_area}.")
```
